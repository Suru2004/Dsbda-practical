```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
dataset = sns.load_dataset('titanic')
dataset.head()
```

```python
numeric_data = dataset.select_dtypes(include=[float, int])
corr = numeric_data.corr()
import seaborn as sns
sns.heatmap(corr)
```

```python
numeric_data = dataset.select_dtypes(include=[float, int])
corr = numeric_data.corr()
import seaborn as sns
sns.heatmap(corr, annot=True)
```

```python
numeric_data = dataset.select_dtypes(include=[float, int])
corr = numeric_data.corr()
sns.heatmap(corr)
```
