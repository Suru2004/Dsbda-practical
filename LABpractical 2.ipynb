```python
Detecting outliers using Z-Score:
```

```python
import numpy as np 
from scipy import stats
```

```python
z_sepal_length = np.abs(stats.zscore(df['sepal_length']))
z_petal_length = np.abs(stats.zscore(df['petal_length']))

print(z_sepal_length)
print(z_petal_length)
```

```python
threshold = 0.18 
z_sepal_length = np.abs(stats.zscore(df['sepal_length']))
z_petal_length = np.abs(stats.zscore(df['petal_length']))

outliers_sepal_length = np.where(z_sepal_length > threshold)
outliers_petal_length = np.where(z_petal_length > threshold)

print(outliers_sepal_length)
print(outliers_petal_length)
```

```python
detecting outliers IQR (Inter Quartile Range)
```

```python
sorted_sepal_length = sorted(df['sepal_length'])
sorted_sepal_length
```

```python
q1 = np.percentile(sorted_sepal_length, 25) 
q3 = np.percentile(sorted_sepal_length, 75) 
print(q1,q3) 
```

```python
IQR = q3-q1 
```

```python
lwr_bound = q1-(1.5*IQR) 
upr_bound = q3+(1.5*IQR) 
print(lwr_bound, upr_bound) 
```

```python
r_outliers = [] 
for i in sorted_sepal_length: 
    if i < lwr_bound or i > upr_bound: 
        r_outliers.append(i)

print(r_outliers)
```

```python
q1 = np.percentile(sorted_sepal_length, 25)
q3 = np.percentile(sorted_sepal_length, 75)
IQR = q3 - q1

lwr_bound = q1 - 1.5 * IQR
upr_bound = q3 + 1.5 * IQR

print(lwr_bound, upr_bound)

r_outliers = [] 
for i in sorted_sepal_length: 
    if i < lwr_bound or i > upr_bound: 
        r_outliers.append(i)

print(r_outliers)
```
