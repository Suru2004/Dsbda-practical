```python
import seaborn as sns
import pandas as pd

# Load the Iris dataset directly from seaborn
iris = sns.load_dataset('iris')

# Display the first few rows of the dataset
iris.head()
```

```python
import matplotlib.pyplot as plt

# Plotting histograms for each feature in the dataset
iris.drop('species', axis=1).hist(figsize=(12, 8), bins=20, edgecolor='black')

# Add titles and labels
plt.suptitle('Feature Distributions in the Iris Dataset', fontsize=16)
plt.show()
```

```python
# Creating box plots for each feature in the dataset
plt.figure(figsize=(12, 8))

# Loop through the numerical columns and create a box plot for each one
for i, column in enumerate(iris.drop('species', axis=1).columns, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(data=iris, x='species', y=column)
    plt.title(f'Boxplot for {column}')

plt.tight_layout()
plt.show()
```
